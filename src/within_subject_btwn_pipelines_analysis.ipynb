```python
from scipy.stats import ttest_rel, wilcoxon

a = df['fwhm5_hrfno_param6']
b = df['fwhm8_hrfyes_param0']
diff = a - b

# Check normality
import scipy.stats as stats
print("Shapiro-Wilk p=", stats.shapiro(diff).pvalue)

if stats.shapiro(diff).pvalue > 0.05:
    tval, pval = ttest_rel(a, b)
    print(f"Paired t-test: t={tval:.3f}, p={pval:.3g}")
else:
    stat, pval = wilcoxon(a, b)
    print(f"Wilcoxon test: stat={stat:.3f}, p={pval:.3g}")
```

```python
import numpy as np

# Pick two pipelines to compare
a = df['fwhm8_hrfyes_param6'].values
b = df['fwhm8_hrfyes_param0'].values

# Observed difference
diff = a - b
observed_mean = np.mean(diff)

# Permutation test
n_permutations = 10000
perm_diffs = np.zeros(n_permutations)

rng = np.random.default_rng(42)

for i in range(n_permutations):
    # Randomly flip the sign of differences
    signs = rng.choice([-1, 1], size=len(diff))
    perm_diffs[i] = np.mean(signs * diff)

# Two-tailed p-value
pval = np.mean(np.abs(perm_diffs) >= np.abs(observed_mean))

print(f"Observed mean difference = {observed_mean:.4f}")
print(f"Permutation p-value = {pval:.5f}")
```

```python
import numpy as np
import pandas as pd

n_permutations = 10000
rng = np.random.default_rng(42)

pvals = {}
observed_diffs = {}

for roi, row in df.iterrows():
    values = row.values  # t-values for each pipeline in this ROI
    k = len(values)
    observed_var = np.var(values)

    perm_vars = np.zeros(n_permutations)
    for i in range(n_permutations):
        perm_vars[i] = np.var(rng.permutation(values))
    
    pval = np.mean(perm_vars >= observed_var)
    pvals[roi] = pval
    observed_diffs[roi] = observed_var

results = pd.DataFrame({'ROI': list(pvals.keys()), 
                        'Observed variance': list(observed_diffs.values()),
                        'Permutation p': list(pvals.values())})
```
